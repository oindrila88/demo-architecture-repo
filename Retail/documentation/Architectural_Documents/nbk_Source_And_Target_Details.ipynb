# Gold Layer Star Schema
* <b>1</b>. <b>dim_customer</b> (<b>DIMENSION</b>)
    * customer_id (<b>PK</b>)
    * customer_name
    * customer_type
    * city
    * state
    * country
    * industry
* <b>2</b>. <b>dim_product</b> (<b>DIMENSION</b>)
    * product_id (<b>PK</b>)
    * product_name
    * category
    * sub_category
    * brand
    * product_segment
    * unit_price
    * supplier_name
* <b>3</b>. <b>dim_calendar</b> (<b>DIMENSION</b>)
    * date_key (<b>PK</b>)
    * full_date
    * day
    * day_name
* <b>4</b>. <b>fact_sales</b> (<b>FACT</b>)
    * transaction_id (<b>PK</b>)
    * date_key (<b>FK</b>)
    * customer_id (<b>FK</b>)
    * product_id (<b>FK</b>)
    * quantity
    * selling_price
    * discount_amount
    * net_sales_amount
    * payment_mode
    * sales_channel
