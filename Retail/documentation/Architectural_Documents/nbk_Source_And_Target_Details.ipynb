# Data Sources
* <b>1</b>. <b>Salesforce CRM</b>
    * Account
    * Opportunity
* <b>2</b>. <b>PostgreSQL Database</b>
    * Product Catalog
    * Inventory
* <b>3</b>. <b>Cloud Storage</b> (<b>CSV</b>)
    * Transaction

# Gold Layer Star Schema
* <b>1</b>. <b>dim_customer</b> (<b>DIMENSION</b>) = <b>salesforce -> account</b>
    * customer_id (<b>PK</b>)
    * customer_name
    * customer_type
    * city
    * state
    * country
    * industry
* <b>2</b>. <b>dim_product</b> (<b>DIMENSION</b>) = <b>postgreSql -> product_catalog</b>
    * product_id (<b>PK</b>)
    * product_name
    * category
    * sub_category
    * brand
    * product_segment
    * unit_price
    * supplier_name
* <b>3</b>. <b>dim_calendar</b> (<b>DIMENSION</b>)
    * date_key (<b>PK</b>)
    * full_date
    * day
    * day_name
* <b>4</b>. <b>fact_sales</b> (<b>FACT</b>) = join of <b>salesforce -> opportunity</b> and <b>blob -> transaction</b>
    * transaction_id (<b>PK</b>)
    * date_key (<b>FK</b>)
    * customer_id (<b>FK</b>)
    * product_id (<b>FK</b>)
    * quantity
    * selling_price
    * discount_amount
    * net_sales_amount
    * payment_mode
    * sales_channel

* Create a metric view definition named <b>retail_metric</b> in the <b>retail_semantic</b> schema. The metric view editor is now open with the following configuration:

Joins:

Customer dimension via account_id
Product dimension via product_id
Calendar dimension via transaction_date
Dimensions (5):

Transaction Date - Date of the sale with date formatting
Year - Transaction year from calendar
Month - Month name for easier analysis
Product Category - High-level product grouping (Electronics, Fashion, Groceries)
Sales Channel - Online vs Store purchases
Measures (5):

Total Revenue - Sum of net sales amount (formatted as USD currency)
Transaction Count - Total number of transactions
Average Order Value - Revenue per transaction (composed from other measures)
Total Quantity - Units sold across all transactions
Average Discount - Mean discount percentage applied
All measures include proper formatting (currency, number, percentage) and synonyms for better discoverability in AI tools. You can now save this as retail_metric in the metric view editor.
